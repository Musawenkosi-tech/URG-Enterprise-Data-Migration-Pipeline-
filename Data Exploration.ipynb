```python

import pandas as pd

df_customers = pd.read_csv("/Volumes/workspace/migration/urg_migration/customers_legacy_large.csv")
df_customers.head(5)

df_orders = pd.read_csv("/Volumes/workspace/migration/urg_migration/orders_legacy_large.csv")
df_orders.head(5)

df_products = pd.read_csv("/Volumes/workspace/migration/urg_migration/products_legacy_large.csv")
df_products.head(5)

df_transactions = pd.read_csv("/Volumes/workspace/migration/urg_migration/transactions_legacy_large.csv")
df_transactions.head(5)


```

```python
df_customers.info()
print(df_customers.isnull().sum())
print(df_customers.duplicated().sum())
duplicated_id = df_customers['customer_id'].astype(str).str.strip()
print(f"Duplicate customer_id: {duplicated_id.duplicated().sum()}")
print(f'Missing customer_id: {df_customers["customer_id"].isnull().sum()}')
print(f'distinct customer_id:{df_customers['customer_id'].nunique()}')


```

```python
df_orders.info()
print(df_orders.isnull().sum())
print(df_orders.duplicated().sum())
duplicated_id = df_orders['order_id'].astype(str).str.strip()
print(f"Duplicate order_id: {duplicated_id.duplicated().sum()}")
print(f'Missing order_id: {df_orders["order_id"].isnull().sum()}')
print(f'distinct order_id:{df_orders['order_id'].nunique()}')

```

```python
df_products.info()
print(df_products.isnull().sum())
print(df_products.duplicated().sum())
duplicated_id = df_products['product_id'].astype(str).str.strip()
print(f"Duplicate product_id: {duplicated_id.duplicated().sum()}")
print(f'Missing product_id: {df_products["product_id"].isnull().sum()}')
print(f'distinct product_id:{df_products['product_id'].nunique()}')
```
